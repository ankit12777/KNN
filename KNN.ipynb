Question 1 : What is the fundamental idea behind ensemble techniques? How does
bagging differ from boosting in terms of approach and objective?

Ensemble idea: Combine multiple models to get better accuracy than a single model.

Bagging:

Builds models independently (parallel)
Uses random samples (bootstrap)
Goal: reduce variance (overfitting)
Example: Random Forest

Boosting:

Builds models sequentially
Each model fixes previous errors
Goal: reduce bias and improve accuracy
Example: AdaBoost, XGBoost

Question 2: Explain how the Random Forest Classifier reduces overfitting compared to
a single decision tree. Mention the role of two key hyperparameters in this process.

Random Forest reduces overfitting by:

Training many decision trees on random samples of data + features
Averaging/voting results → reduces variance of a single tree

Why better than one tree:

Single tree memorizes data → overfits
Random Forest adds randomness → makes model more general

Question 3: What is Stacking in ensemble learning? How does it differ from traditional
bagging/boosting methods? Provide a simple example use case.


Stacking:

Combines multiple different models (base learners)
A meta-model learns from their outputs to make final prediction

Difference:

Bagging: parallel, same type models, average/vote
Boosting: sequential, fixes errors
Stacking: uses different models + meta learner

Example use case:

Predicting house prices using:
Linear Regression + Decision Tree + XGBoost.

Question 4:What is the OOB Score in Random Forest, and why is it useful? How does
it help in model evaluation without a separate validation set?

OOB (Out-of-Bag) Score:

Accuracy calculated using data not used in training each tree

Why useful:

Gives built-in validation for Random Forest
No need for separate validation set

How it works:

Each tree uses bootstrap samples (some data left out)
Left-out data = OOB data
Model is tested on these → average accuracy = OOB score.

AdaBoost:

Errors: Focuses on wrongly classified points
Weights: Increases weight of misclassified samples
Use cases: Simple classification problems, less noisy data

Gradient Boosting:

Errors: Fits new model on residual errors
Weights: No direct sample weighting; uses gradient of loss function
Use cases: Complex tasks (regression, classification, Kaggle models).

Question 6:Why does CatBoost perform well on categorical features without requiring
extensive preprocessing? Briefly explain its handling of categorical variables?


CatBoost:

Handles categorical data internally (no heavy encoding needed)

How it works:

Uses target-based encoding with ordered statistics
Converts categories into meaningful numerical values safely
Reduces data leakage using ordered boosting

Why it performs well:

No need for manual one-hot/label encoding
Handles high-cardinality categories efficiently
More stable and less overfitting on categorical features.


Question 7: KNN Classifier Assignment: Wine Dataset Analysis with
Optimization
Task:
1. Load the Wine dataset (sklearn.datasets.load_wine()).
2. Split data into 70% train and 30% test.
3. Train a KNN classifier (default K=5) without scaling and evaluate using:
a. Accuracy
b. Precision, Recall, F1-Score (print classification report)
4. Apply StandardScaler, retrain KNN, and compare metrics.
5. Use GridSearchCV to find the best K (test K=1 to 20) and distance metric
(Euclidean, Manhattan).
6. Train the optimized KNN and compare results with the unscaled/scaled versions.

```python
from sklearn.datasets import load_wine
from sklearn.model_selection import train_test_split, GridSearchCV
from sklearn.neighbors import KNeighborsClassifier
from sklearn.preprocessing import StandardScaler
from sklearn.metrics import classification_report, accuracy_score

# 1. Load data
data = load_wine()
X, y = data.data, data.target

# 2. Split data
X_train, X_test, y_train, y_test = train_test_split(
    X, y, test_size=0.3, random_state=42
)

# 3. KNN without scaling
knn = KNeighborsClassifier(n_neighbors=5)
knn.fit(X_train, y_train)
pred1 = knn.predict(X_test)

print("Without Scaling Accuracy:", accuracy_score(y_test, pred1))
print(classification_report(y_test, pred1))

# 4. With scaling
scaler = StandardScaler()
X_train_s = scaler.fit_transform(X_train)
X_test_s = scaler.transform(X_test)

knn.fit(X_train_s, y_train)
pred2 = knn.predict(X_test_s)

print("With Scaling Accuracy:", accuracy_score(y_test, pred2))
print(classification_report(y_test, pred2))

# 5. GridSearchCV for best K and metric
param_grid = {
    'n_neighbors': list(range(1, 21)),
    'metric': ['euclidean', 'manhattan']
}

grid = GridSearchCV(KNeighborsClassifier(), param_grid, cv=5)
grid.fit(X_train_s, y_train)

print("Best Params:", grid.best_params_)

# 6. Best model
best_knn = grid.best_estimator_
pred3 = best_knn.predict(X_test_s)

print("Optimized KNN Accuracy:", accuracy_score(y_test, pred3))
print(classification_report(y_test, pred3))
```

Question 8 : PCA + KNN with Variance Analysis and Visualization
Task:
1. Load the Breast Cancer dataset (sklearn.datasets.load_breast_cancer()).
2. Apply PCA and plot the scree plot (explained variance ratio).
3. Retain 95% variance and transform the dataset.
4. Train KNN on the original data and PCA-transformed data, then compare
accuracy.
5. Visualize the first two principal components using a scatter plot (color by class).

```python
from sklearn.datasets import load_breast_cancer
from sklearn.model_selection import train_test_split
from sklearn.preprocessing import StandardScaler
from sklearn.decomposition import PCA
from sklearn.neighbors import KNeighborsClassifier
from sklearn.metrics import accuracy_score
import matplotlib.pyplot as plt
import numpy as np

# 1. Load data
data = load_breast_cancer()
X, y = data.data, data.target

# Split
X_train, X_test, y_train, y_test = train_test_split(
    X, y, test_size=0.3, random_state=42
)

# Scale
scaler = StandardScaler()
X_train_s = scaler.fit_transform(X_train)
X_test_s = scaler.transform(X_test)

# 2. PCA for scree plot
pca_full = PCA().fit(X_train_s)

plt.plot(np.cumsum(pca_full.explained_variance_ratio_))
plt.xlabel("Components")
plt.ylabel("Explained Variance")
plt.title("Scree Plot")
plt.show()

# 3. 95% variance PCA
pca = PCA(0.95)
X_train_pca = pca.fit_transform(X_train_s)
X_test_pca = pca.transform(X_test_s)

print("Components kept:", pca.n_components_)

# 4. KNN on original scaled data
knn1 = KNeighborsClassifier()
knn1.fit(X_train_s, y_train)
pred1 = knn1.predict(X_test_s)

print("Original Accuracy:", accuracy_score(y_test, pred1))

# KNN on PCA data
knn2 = KNeighborsClassifier()
knn2.fit(X_train_pca, y_train)
pred2 = knn2.predict(X_test_pca)

print("PCA Accuracy:", accuracy_score(y_test, pred2))

# 5. 2D visualization
pca2 = PCA(n_components=2)
X_2d = pca2.fit_transform(X_train_s)

plt.scatter(X_2d[:,0], X_2d[:,1], c=y_train, cmap='viridis')
plt.xlabel("PC1")
plt.ylabel("PC2")
plt.title("PCA (2 Components)")
plt.show()
```

Question 9:KNN Regressor with Distance Metrics and K-Value
Analysis
Task:
1. Generate a synthetic regression dataset
(sklearn.datasets.make_regression(n_samples=500, n_features=10)).
2. Train a KNN regressor with:
a. Euclidean distance (K=5)
b. Manhattan distance (K=5)
c. Compare Mean Squared Error (MSE) for both.
3. Test K=1, 5, 10, 20, 50 and plot K vs. MSE to analyze bias-variance tradeoff

```python
from sklearn.datasets import make_regression
from sklearn.model_selection import train_test_split
from sklearn.neighbors import KNeighborsRegressor
from sklearn.metrics import mean_squared_error
import matplotlib.pyplot as plt

# 1. Data
X, y = make_regression(n_samples=500, n_features=10, noise=10, random_state=42)

X_train, X_test, y_train, y_test = train_test_split(
    X, y, test_size=0.3, random_state=42
)

# 2a. Euclidean (K=5)
knn_eu = KNeighborsRegressor(n_neighbors=5, metric='euclidean')
knn_eu.fit(X_train, y_train)
pred_eu = knn_eu.predict(X_test)
mse_eu = mean_squared_error(y_test, pred_eu)

# 2b. Manhattan (K=5)
knn_man = KNeighborsRegressor(n_neighbors=5, metric='manhattan')
knn_man.fit(X_train, y_train)
pred_man = knn_man.predict(X_test)
mse_man = mean_squared_error(y_test, pred_man)

print("MSE Euclidean:", mse_eu)
print("MSE Manhattan:", mse_man)

# 3. K vs MSE
ks = [1, 5, 10, 20, 50]
mse_list = []

for k in ks:
    knn = KNeighborsRegressor(n_neighbors=k)
    knn.fit(X_train, y_train)
    pred = knn.predict(X_test)
    mse_list.append(mean_squared_error(y_test, pred))

plt.plot(ks, mse_list, marker='o')
plt.xlabel("K Value")
plt.ylabel("MSE")
plt.title("K vs MSE (Bias-Variance Tradeoff)")
plt.show()
```
